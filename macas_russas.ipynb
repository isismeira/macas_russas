*Calculando o coeficiente angular (a)*

```python
Y = Moscow
X = datas
n = np.size(Moscow)
```

```python
(X**2).shape
```

```python
a = (n*np.sum(X*Y) - np.sum(X)*np.sum(Y))/(n*np.sum(X**2)- np.sum(X)**2)
```

*Calculando o coeficiente linear (b)*

```python
b = np.mean(Y) - a*np.mean(X)
```

*Agora podemos calcular y*

```python
y = a*X + b
```

```python
np.linalg.norm(Moscow-y)
```

Todo processo que realizamos é uma *regressão linear*

Novo gráfico com reta ajustada

```python
plt.plot(datas, Moscow)
plt.plot(x,y)
plt.plot(41.5, 41.5*a+b)
plt.plot(100, 100*a+b, '*r' ) # estimando valores no futuro
```
